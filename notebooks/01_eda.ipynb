```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Cargar los datos desde la carpeta 'data'
# Usamos '../' para salir de la carpeta notebooks y entrar a data
df = pd.read_csv('../data/winequality-red.csv')

# 2. Ver las primeras 5 filas para entender qué variables tenemos
print("--- Primeras filas del dataset ---")
display(df.head())

# 3. Ver información general (tipos de datos y si hay valores nulos/vacíos)
print("\n--- Información del dataset ---")
df.info()

# 4. Ver estadísticas básicas (promedios, mínimos, máximos)
print("\n--- Estadísticas descriptivas ---")
display(df.describe())

# 5. Graficar la variable que queremos predecir (quality)
plt.figure(figsize=(8, 5))
sns.countplot(x='quality', data=df, palette='viridis')
plt.title('Distribución de la Calidad del Vino Tinto')
plt.xlabel('Calidad (Puntuación)')
plt.ylabel('Cantidad de Vinos')
plt.show()
```
